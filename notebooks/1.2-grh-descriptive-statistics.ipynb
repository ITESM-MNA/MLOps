# IQR

$$
IQR = Q_3 - Q_1
$$

```python
IQR = df["fixed acidity"].quantile(0.75) - df["fixed acidity"].quantile(0.25)
print("IQR: ", IQR)
```

$$
lower = Q_1 - 1.5 * IQR
$$

$$
upper = Q_3 + 1.5 * IQR
$$

```python
lower = df["fixed acidity"].quantile(0.25) - 1.5 * IQR
upper = df["fixed acidity"].quantile(0.75) + 1.5 * IQR

print(lower, upper)
```

```python
outliers = df[(df['fixed acidity'] < lower) | (df['fixed acidity'] > upper)]
print(f"Potential outliers in fixed acidity: {len(outliers):,}")
```

```python
df.reset_index(inplace=True)
df['quality'] = df['quality'].astype('category')

df.head()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='quality', y='fixed acidity', data=df)
plt.title('Boxplot of Fixed Acidity by Quality')
plt.xlabel('Quality')
plt.ylabel('Fixed Acidity')
plt.show()
```
